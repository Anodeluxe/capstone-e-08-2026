# Statistika Deskriptif Dataset Mentah (watera.csv)

Notebook ini menghitung statistika deskriptif dari dataset mentah `./data/raw/watera.csv`,
dengan fokus pada kolom **ph**, **tds**, dan **turbidity**.

Nilai yang dihitung meliputi min, max, mean, median, std, serta jumlah data kosong (missing).

```python
import pandas as pd

pd.set_option('display.float_format', lambda x: f'{x:,.2f}')

df = pd.read_csv('./data/raw/watera.csv')
print(f'Jumlah baris: {len(df)} | Jumlah kolom: {len(df.columns)}')
print(f'Kolom: {list(df.columns)}')
df.head()
```

## Statistika Deskriptif Kolom ph, tds, turbidity

```python
kolom = ['ph', 'tds', 'turbidity']

stats = pd.DataFrame({
    'min': df[kolom].min(),
    'max': df[kolom].max(),
    'mean': df[kolom].mean(),
    'median': df[kolom].median(),
    'std': df[kolom].std(),
    'missing': df[kolom].isna().sum(),
})

stats
```

## Ringkasan Min / Max

```python
for col in kolom:
    print(f'{col:>10} -> min: {df[col].min():.2f} | max: {df[col].max():.2f}')
```

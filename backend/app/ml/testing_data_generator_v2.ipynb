```python
import pandas as pd

print("Membuat dataset pengujian khusus untuk simulasi...")

# 1. Baca data yang sudah diproses (1000 siklus)
df_full = pd.read_csv('./data/processed/data_toren_v2.csv')

# 2. Ambil satu siklus toren secara acak (misalnya toren urutan terakhir)
siklus_tes = df_full['Cycle_ID'].max()
df_testing = df_full[df_full['Cycle_ID'] == siklus_tes].copy()

# 3. Urutkan berdasarkan Hari dan reset index agar berurutan (0, 1, 2, dst)
df_testing = df_testing.sort_values(by='Hari', ascending=True)
df_testing.reset_index(drop=True, inplace=True)

# 4. Simpan menjadi file baru
# df_scaled.to_csv('data/processed/data_toren_v2.csv', index=False)
nama_file_testing = 'data_testing_v2.csv'
df_testing.to_csv('data/testing/data_testing_v2.csv', index=False)

print(f"Selesai! Toren ID-{siklus_tes} terpilih sebagai bahan ujian.")
print(f"Total umur toren ini adalah: {len(df_testing)} hari.")
print(f"File berhasil disimpan sebagai: '{nama_file_testing}'")
```
